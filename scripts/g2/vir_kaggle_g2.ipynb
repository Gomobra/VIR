```python
import subprocess, json, time, os
log = open('/kaggle/working/log.txt','w')
def P(*a):
    s=' '.join(str(x) for x in a)
    print(s); log.write(s+'\n'); log.flush()
t0=time.time()
r=subprocess.run(['git','clone','--depth','1','https://github.com/Gomobra/VIR.git','/tmp/VIR'], capture_output=True, text=True)
P('clone rc', r.returncode)
env={**os.environ,'CARGO_TARGET_DIR':'/tmp/target'}
b=subprocess.run(['cargo','build','--release','--manifest-path','/tmp/VIR/crates/vir-vq/Cargo.toml','--features','cuda'], capture_output=True, text=True, env=env)
P('build rc', b.returncode); P(b.stderr[-3000:])
t=subprocess.run(['cargo','test','--release','--manifest-path','/tmp/VIR/crates/vir-vq/Cargo.toml','--features','cuda','--','--nocapture'], capture_output=True, text=True, env=env)
P('test rc', t.returncode); P(t.stdout[-1500:]); P(t.stderr[-1500:])
result={'build_rc':b.returncode,'test_rc':t.returncode,'elapsed_s':time.time()-t0}
P('RESULT', result)
open('/kaggle/working/result.json','w').write(json.dumps(result)); log.close()
```
