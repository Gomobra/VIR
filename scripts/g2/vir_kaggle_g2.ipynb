# VIR Vbeta - G2 CUDA smoke
Build Burn dengan backend CUDA dari source (di /tmp, bukan /kaggle/working), jalankan smoke di GPU, tulis hasil ke result.json.

```python
!curl --proto '=https' --tlsv1.2 -sSf https://sh.rustup.rs | sh -s -- -y --default-toolchain stable
import os
os.environ['PATH'] = os.path.expanduser('~/.cargo/bin') + ':' + os.environ['PATH']
```

```python
import subprocess, json, time
t0 = time.time()
r = subprocess.run(['git','clone','--depth','1','https://github.com/Gomobra/VIR.git','/tmp/VIR'], capture_output=True, text=True)
print('clone rc', r.returncode)
build = subprocess.run(['cargo','build','--release','--manifest-path','/tmp/VIR/crates/vir-vq/Cargo.toml','--features','cuda'], capture_output=True, text=True, env={**os.environ,'CARGO_TARGET_DIR':'/tmp/target'})
print('build rc', build.returncode)
print(build.stderr[-2000:])
test = subprocess.run(['cargo','test','--release','--manifest-path','/tmp/VIR/crates/vir-vq/Cargo.toml','--features','cuda','--','--nocapture'], capture_output=True, text=True, env={**os.environ,'CARGO_TARGET_DIR':'/tmp/target'})
print('test rc', test.returncode)
print(test.stdout[-1000:])
print(test.stderr[-1000:])
result = {'build_rc': build.returncode, 'test_rc': test.returncode, 'elapsed_s': time.time()-t0}
open('/kaggle/working/result.json','w').write(json.dumps(result))
print('RESULT', result)
```
