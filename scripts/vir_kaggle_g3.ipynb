```python
import subprocess, json, time, os
log = open('/kaggle/working/log.txt','w')
def P(*a):
    s=' '.join(str(x) for x in a)
    print(s); log.write(s+'\n'); log.flush()
t0=time.time()
subprocess.run(['git','clone','--depth','1','https://github.com/Gomobra/VIR.git','/tmp/VIR'], check=True)
env={**os.environ,'CARGO_TARGET_DIR':'/tmp/target'}
b=subprocess.run(['cargo','build','--release','--manifest-path','/tmp/VIR/crates/vir-train/Cargo.toml','--features','cuda'], capture_output=True, text=True, env=env)
P('build rc', b.returncode); P(b.stderr[-3000:])
t=subprocess.run(['cargo','test','--release','--manifest-path','/tmp/VIR/crates/vir-train/Cargo.toml','--features','cuda','--','--nocapture'], capture_output=True, text=True, env=env)
P('test rc', t.returncode); P(t.stdout[-2000:]); P(t.stderr[-1500:])
P('RESULT', {'build_rc':b.returncode,'test_rc':t.returncode,'elapsed':time.time()-t0})
open('/kaggle/working/result.json','w').write(json.dumps({'build_rc':b.returncode,'test_rc':t.returncode}))
log.close()
```
